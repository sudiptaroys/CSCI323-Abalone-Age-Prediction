# MLP — Abalone age prediction

The network has 10 inputs, hidden layers of 64 and 32 ReLU units, and one linear regression output. It minimizes squared error with Adam using train-scaled numerical inputs. Settings: max_iter=1000, random_state=42, alpha=0.0001, learning_rate_init=0.001, batch_size=auto, shuffle=True, early_stopping=False. A convergence warning is retained if raised.


```python
import os
for key in ['OMP_NUM_THREADS', 'OPENBLAS_NUM_THREADS', 'MKL_NUM_THREADS']:
    os.environ.setdefault(key, '1')
from pathlib import Path
import pandas as pd
from abalone_project import prepare_data, load_data, fit_method, subgroup_comparison
if not Path('X_train_scaled.csv').exists():
    prepare_data()
data = load_data()
```

The network has 10 inputs, hidden layers of 64 and 32 ReLU units, and one linear regression output. It minimizes squared error with Adam using train-scaled numerical inputs. Settings: max_iter=1000, random_state=42, alpha=0.0001, learning_rate_init=0.001, batch_size=auto, shuffle=True, early_stopping=False. A convergence warning is retained if raised.

```python
model, predictions, scores = fit_method('MLP', data)
display(pd.DataFrame([scores]))
```

Compare models trained on all training rows (global) against models trained separately on F/I/M rows (local), using the same test subsets. The historical RF comparison changes both parameters and training population, so its interpretation is exploratory. XGBoost and MLP clone the global parameters. Smaller local datasets can increase variance; specialization does not guarantee improvement.

```python
subgroups = subgroup_comparison('MLP', model, data)
display(subgroups)
```

```python
import matplotlib.pyplot as plt
fig, ax = plt.subplots(figsize=(6,4))
ax.scatter(data['y_test'], predictions, s=12, alpha=.4)
limits = [data['y_test'].min(), data['y_test'].max()]
ax.plot(limits, limits, 'r--')
ax.set(xlabel='Actual age (years)', ylabel='Predicted age (years)')
display(fig)
plt.close(fig)
```

MAE measures typical absolute error in years; MSE penalizes large errors more strongly; R² compares squared error with a mean-target baseline. One test split cannot establish statistical superiority.
